```python
caminho_ficheiro = 'Tabela_13_Meio_de_transporte.xlsx'
df = pd.read_excel(
    caminho_ficheiro, sheet_name='Exemplo gráfico Brasil', skiprows=1
)
df.columns = ['Meio de Transporte', 'Branca', 'Preta ou Parda', 'Indígena']
```

```python
df
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Carregar os dados do ficheiro Excel


# 2. Configurar os gráficos
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
cores = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#c2c2f0', '#ffb3e6']

grupos = ['Branca', 'Preta ou Parda', 'Indígena']

for i, col in enumerate(grupos):
  axes[i].pie(
      df[col],
      labels=df['Meio de Transporte'],
      autopct='%1.2f%%',  # Mostra a percentagem com 1 casa decimal
      startangle=180,  # Ângulo inicial de rotação
      colors=cores,
      wedgeprops={'edgecolor': 'white', 'linewidth': 1.2},
  )
  axes[i].set_title(f'Cor/Raça: {col}', fontsize=12, fontweight='bold')

plt.suptitle(
    'Distribuição Percentual por Meio de Transporte (Brasil - 2022)',
    fontsize=14,
    fontweight='bold',
)
plt.tight_layout()
plt.savefig('grafico_transporte.png', dpi=300)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Dados
categorias = ['Carro', 'Autocarro', 'A pé', 'Bicicleta', 'Outros']
valores = [40, 25, 20, 10, 5]
cores = ['#3498db', '#e74c3c', '#2ecc71', '#f1c40f', '#9b59b6']
destaque = (0.1, 0, 0, 0, 0)  # Destaca a primeira fatia (Carro)

# Criar o gráfico
plt.figure(figsize=(7, 7))
plt.pie(
    valores,
    labels=categorias,
    autopct='%1.1f%%',
    startangle=90,
    colors=cores,
    explode=destaque,  # Afasta a fatia selecionada
    shadow=True,  # Adiciona sombra ao gráfico
)

plt.title('Distribuição de Meios de Transporte', fontsize=14)
plt.show()
```
